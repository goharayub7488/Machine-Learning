```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
URL ="https://www.timeanddate.com/weather/?low=c"
webPage=requests.get(URL).text
```

```python
from decorator import append
soup=BeautifulSoup(webPage,'lxml')
cities=soup.find_all('tr')
name=[]
dayTime=[]
temperature=[]
for i in cities:
  links=i.find_all('a')
  for link in links:
    name.append(link.text)
  DayTimeCell=i.find_all('td',class_='r')
  for day in DayTimeCell:
    if day.text=="":
      continue
    else:
      dayTime.append(day.text)
  temp=i.find_all('td',class_='rbi')
  for t in temp:
    temperature.append(t.text)

```

```python
print(len(name))
print(len(dayTime))
print(len(temperature))
```

```python
d={'Cities':name,'DayTime':dayTime,'Temperature':temperature}
df=pd.DataFrame(d)
df
```

```python
df = df.sort_values(by='Cities').reset_index(drop=True)
```
